# List of largest companies in the United States by revenue

### Importing libraries

```python
import requests
import pandas as pd
from bs4 import BeautifulSoup
import os
```

### Initialization of known entities

```python
url = 'https://en.wikipedia.org/wiki/List_of_largest_companies_in_the_United_States_by_revenue'
csv_path = 'data/us_companies_by_revenue.csv'

```

### Loading the web page for webscrapping

```python
headers = {"User-Agent": "Mozilla/5.0"}
page = requests.get(url, headers=headers)

soup = BeautifulSoup(page.content, "html.parser")
```

### Srapping of required information

```python
tables = soup.find_all('table')
table = tables[0] # the first table contains the data we need

rows = table.find_all('tr')

data = []

for row in rows:
    cols = row.find_all('td')
    individual_row_data = [col.text.strip() for col in cols]

    if individual_row_data: # skip empty rows
        data.append(individual_row_data)

df = pd.DataFrame(data, columns=["Rank", "Name", "Industry", "Revenue (USD millions)", "Revenue growth", "Employees", "Headquarters"])

print(df.head())
```

```python
os.makedirs("data", exist_ok=True)

df.to_csv(csv_path, index=False)
```
